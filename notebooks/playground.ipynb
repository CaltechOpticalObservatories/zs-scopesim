```python
%matplotlib inline
import os
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm
from astropy import units as u

import scopesim as sim
import scopesim_templates as sim_tp
```

```python
sim.download_packages(["OSIRIS","LaPalma","GTC"])
```

```python
## Set up OSIRIS
cmd = sim.UserCommands(use_instrument="OSIRIS", set_modes=["MAAT"])
inst = sim.OpticalTrain(cmd)
```

```python
inst.effects
```

```python
inst["maat_image_slicer"].data
```

```python
source = sim_tp.stellar.star_grid(n=1, mmin=11, mmax=11.001, separation=1)
inst.observe(source, update=True)
```

```python
inst.optics_manager.fov_effects
```

```python
plt.figure(figsize=(8,8))
plt.imshow(inst.image_planes[0].data, origin="lower",cmap="viridis")
plt.colorbar()
```

```python
hdus = inst.readout()
plt.figure(figsize=(8,8))
plt.imshow(hdus[0][1].data, origin="lower", cmap="Grays")
plt.colorbar()
```
