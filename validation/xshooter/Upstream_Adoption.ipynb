# Bounded observations after upstream adoption

These observations exercise the adopted engine beyond X-shooter and ZShooter. Narrow wavelength intervals bound cube work while retaining optics, PSFs and native detector geometry. MICADO spectroscopy uses the full focal plane because the selected orders miss its default central detector window.

Each case requires positive source flux, finite image planes and every populated science extension, plus INSTRUME, DIT, NDIT and EXPTIME metadata. Any logged readout error fails the run. The earlier HAWKI mode/header traceback was an actual failure: its missing default mode list is now repaired. Large rate arrays are saved locally; the compact results table is committed.


```python
from pathlib import Path
import sys
import gc
import pandas as pd
import scopesim as sim
from zs_scopesim_tools.helpers import configure_irdb_path, disable_scopesim_progress_bars, set_scopesim_log_level

workspace = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'env/zshooter-stack.toml').exists())
sys.path.insert(0, str(workspace / 'validation/xshooter'))
from instrument_probe import observe_case
irdb = Path(configure_irdb_path())
output = workspace / 'validation/xshooter/cache/upstream_adopted'
output.mkdir(parents=True, exist_ok=True)
set_scopesim_log_level('warning')
print(sim.__version__, sim.__file__)
```

```python
cases = [
    ('MICADO_SCAO', 'MICADO', ['SCAO', 'IMG_4mas'], 2.17, 2.18, {'!OBS.filter_name_fw1': 'open', '!OBS.filter_name_fw2': 'Ks'}),
    ('MICADO_MCAO', 'MICADO', ['MCAO', 'IMG_4mas'], 2.17, 2.18, {'!OBS.filter_name_fw1': 'open', '!OBS.filter_name_fw2': 'Ks'}),
    ('MICADO_SPEC', 'MICADO', ['SCAO', 'SPEC'], 1.2, 1.202, {'!OBS.slit': 'Wide', '!OBS.filter_name_fw1': 'Spec_IJ', '!OBS.filter_name_fw2': 'open', '!DET.include_full': True, '!DET.include_window': False}),
    ('METIS_img_lm', 'METIS', ['img_lm'], 3.79, 3.8, {}),
    ('METIS_img_n', 'METIS', ['img_n'], 10.5, 10.51, {}),
    ('METIS_lss_l', 'METIS', ['lss_l'], 3.79, 3.8, {}),
    ('METIS_lss_n', 'METIS', ['lss_n'], 10.5, 10.51, {}),
    ('METIS_lms', 'METIS', ['lms'], 3.498, 3.502, {'!OBS.wavelen': 3.5}),
    ('HAWKI', 'HAWKI', [], 2.17, 2.18, {'!OBS.filter_name': 'Ks'}),
]
rows = []
for case, instrument, modes, lower, upper, properties in cases:
    with disable_scopesim_progress_bars():
        results = observe_case(irdb, output / f'{case}.npz', instrument, modes, lower, upper, properties, check_readout=True)
    for index, shape, background, source in results:
        rows.append((case, index, shape[0], shape[1], background, source))
    print(case, 'finite readout; source e/s', sum(row[-1] for row in results), flush=True)
    gc.collect()
results_table = pd.DataFrame(rows, columns=['instrument_mode', 'image_plane', 'rows', 'columns', 'background_e_s', 'source_e_s'])
results_table.to_csv(workspace / 'validation/xshooter/data/adopted_upstream_observations.csv', index=False)
display(results_table)
```

## MOSAIC metadata check

MOSAIC now supplies its instrument identity in IRDB, and its table output preserves the actual DIT, NDIT and accumulated exposure recorded by the detector integration. The same strict readout check passes below; neither missing metadata nor logged readout errors are ignored. The nine earlier cases, including HAWKI, remain in `data/adopted_upstream_observations.csv`; this case is saved separately in `data/adopted_mosaic_observation.csv`.

```python
with disable_scopesim_progress_bars():
    mosaic_results = observe_case(irdb, output/'MOSAIC_MOS.npz', 'MOSAIC', ['MOS-LR-R'], .75, .751, {}, check_readout=True)
mosaic_rows = [('MOSAIC_MOS', index, shape[0], shape[1], background, source) for index, shape, background, source in mosaic_results]
mosaic_table = pd.DataFrame(mosaic_rows, columns=['instrument_mode', 'image_plane', 'rows', 'columns', 'background_e_s', 'source_e_s'])
mosaic_table.to_csv(workspace/'validation/xshooter/data/adopted_mosaic_observation.csv', index=False)
display(mosaic_table)
```
