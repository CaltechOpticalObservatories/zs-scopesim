# Bounded upstream observations

These observations supplement the upstream test suite, whose MICADO spectroscopy tests are skipped for memory use. They use the real instrument configurations and a narrow wavelength interval. Electron-rate arrays before detector noise separate sky/background and a centered flat-spectrum point source. Compare the saved arrays between the clean upstream revision and the review port; these are numerical regression checks, not absolute instrument calibrations.

```python
import sys
from pathlib import Path

# Checkouts and the isolated upstream installation prepared as described in the validation notes.
review = Path('/tmp/xshooter-review')
engine = 'upstream-port'  # Run 'upstream-scopesim' first; restart the kernel when changing engine.
sys.path[:0] = [str(review / engine), str(review / 'upstream-runtime'), str(Path.cwd())]
from instrument_probe import observe_case
from skycalc_ipy.core import ESOQueryBase
import scopesim as sim

ESOQueryBase.REQUEST_TIMEOUT = 120
irdb = review / 'upstream-irdb'
print(sim.__version__, sim.__file__)
```

```python
results = {}
for ao in ['SCAO', 'MCAO']:
    name = 'MICADO_' + ao
    results[name] = observe_case(irdb, review / f'{engine}_{name}.npz', 'MICADO', [ao, 'IMG_4mas'], 2.17, 2.18, {'!OBS.filter_name_fw1': 'open', '!OBS.filter_name_fw2': 'Ks'})
    print(name, results[name])
```

```python
# At 1.201 um the orders lie outside the central detector window.
# Use the native full array; the narrow wavelength interval bounds cube memory.
name = 'MICADO_SPEC'
results[name] = observe_case(irdb, review / f'{engine}_{name}.npz', 'MICADO', ['SCAO', 'SPEC'], 1.2, 1.202, {'!OBS.slit': 'Wide', '!OBS.filter_name_fw1': 'Spec_IJ', '!OBS.filter_name_fw2': 'open', '!DET.include_full': True, '!DET.include_window': False})
print(name, results[name])
```

```python
for mode, lower, upper in [('img_lm', 3.79, 3.8), ('img_n', 10.5, 10.51), ('lss_l', 3.79, 3.8), ('lss_n', 10.5, 10.51), ('lms', 3.498, 3.502)]:
    name = 'METIS_' + mode
    properties = {'!OBS.wavelen': 3.5} if mode == 'lms' else {}
    results[name] = observe_case(irdb, review / f'{engine}_{name}.npz', 'METIS', [mode], lower, upper, properties)
    print(name, results[name])
```

```python
name = 'MOSAIC_MOS'
results[name] = observe_case(irdb, review / f'{engine}_{name}.npz', 'MOSAIC', ['MOS-LR-R'], .75, .751, {})
print(name, results[name])
name = 'HAWKI'
results[name] = observe_case(irdb, review / f'{engine}_{name}.npz', 'HAWKI', [], 2.17, 2.18, {'!OBS.filter_name': 'Ks'})
print(name, results[name])
```

Compare both engine runs using the saved image planes. These are electrons per second before detector noise. The position and RMS-width differences are in image-plane pixels, not wavelength shifts or fitted resolving powers. An unchanged sum alone does not establish an unchanged spatial profile. See [the validation assessment](../../docs/xshooter-validation.md) for the interpretation and the scope of the port.

```python
from compare_instruments import compare_instruments

comparison = compare_instruments(review)
assert set(comparison.instrument) == set(results)
comparison.to_csv('data/upstream_observations.csv', index=False, float_format='%.8g')
comparison.round(6)
```
